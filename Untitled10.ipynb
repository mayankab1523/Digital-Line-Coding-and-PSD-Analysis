```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# 1. Helper Functions (Quantization & Signals)
# ==========================================
def uniform_quantizer(signal, levels, min_val, max_val):
    """Simple uniform quantizer for PCM/DPCM."""
    step = (max_val - min_val) / levels
    quantized = np.round((signal - min_val) / step) * step + min_val
    # Clip to max/min bounds
    quantized = np.clip(quantized, min_val, max_val)
    return quantized

def generate_signal(f, fs, duration, amplitude=1.0):
    """Generates a sine wave."""
    t = np.arange(0, duration, 1/fs)
    return t, amplitude * np.sin(2 * np.pi * f * t)

# ==========================================
# 2. Modulation Implementations
# ==========================================
def simulate_pcm(signal, bits, max_val):
    levels = 2**bits
    quantized = uniform_quantizer(signal, levels, -max_val, max_val)
    error = signal - quantized
    return quantized, error

def simulate_dpcm(signal, bits, max_val, alpha=0.9):
    """First-order DPCM."""
    N = len(signal)
    levels = 2**bits

    predicted = np.zeros(N)
    reconstructed = np.zeros(N)
    quantized_error = np.zeros(N)
    prediction_error = np.zeros(N)

    for i in range(1, N):
        # 1. Prediction (based on previous reconstructed sample)
        predicted[i] = alpha * reconstructed[i-1]

        # 2. Prediction Error
        prediction_error[i] = signal[i] - predicted[i]

        # 3. Quantization of Error
        quantized_error[i] = uniform_quantizer(prediction_error[i], levels, -max_val, max_val)

        # 4. Reconstruction
        reconstructed[i] = predicted[i] + quantized_error[i]

    return predicted, reconstructed, prediction_error, quantized_error

def simulate_delta_modulation(signal, delta):
    """Delta Modulation (1-bit DPCM)."""
    N = len(signal)
    reconstructed = np.zeros(N)

    for i in range(1, N):
        # Error between current signal and previous staircase value
        error = signal[i] - reconstructed[i-1]

        # Quantize to +Delta or -Delta (1-bit)
        if error >= 0:
            step = delta
        else:
            step = -delta

        # Reconstruction (Staircase)
        reconstructed[i] = reconstructed[i-1] + step

    return reconstructed

# ==========================================
# 3. Running the Simulations & Visualizations
# ==========================================
# Parameters
fs = 1000       # Sampling frequency
duration = 0.05 # 50 ms
f_slow = 20     # Slowly varying input (20 Hz)
f_fast = 150    # Rapidly varying input (150 Hz)

t, sig_slow = generate_signal(f_slow, fs, duration)
t, sig_fast = generate_signal(f_fast, fs, duration)

plt.figure(figsize=(15, 18))

# --- Visualization 1: DPCM vs PCM (Tasks 14 & 15) ---
pcm_recon, pcm_err = simulate_pcm(sig_slow, bits=3, max_val=1.2)
dpcm_pred, dpcm_recon, dpcm_pred_err, dpcm_quant_err = simulate_dpcm(sig_slow, bits=3, max_val=1.2, alpha=1.0)

plt.subplot(4, 1, 1)
plt.plot(t, sig_slow, 'k', label='Original Signal', linewidth=2)
plt.plot(t, dpcm_pred, 'g--', label='DPCM Predicted', alpha=0.7)
plt.plot(t, dpcm_recon, 'b-o', label='DPCM Reconstructed', markersize=4)
plt.title('DPCM: Original, Predicted, and Reconstructed (Slow Signal)')
plt.legend()
plt.grid(True)

plt.subplot(4, 1, 2)
plt.plot(t, pcm_err, 'r', label='PCM Error')
plt.plot(t, dpcm_pred_err, 'b', label='DPCM Prediction Error')
plt.title('Error Comparison: PCM vs DPCM')
plt.legend()
plt.grid(True)

# --- Visualization 2: Delta Modulation Staircase (Tasks 16 & 17) ---
# Granular Noise: Large step size on a slow signal
delta_large = 0.15
dm_granular = simulate_delta_modulation(sig_slow, delta_large)

# Slope Overload: Small step size on a fast signal
delta_small = 0.05
dm_overload = simulate_delta_modulation(sig_fast, delta_small)

plt.subplot(4, 2, 5)
plt.plot(t, sig_slow, 'k', label='Original (Slow)')
plt.step(t, dm_granular, 'r', where='post', label=f'DM Staircase (Δ={delta_large})')
plt.title('Delta Modulation: Granular Noise')
plt.legend()
plt.grid(True)

plt.subplot(4, 2, 6)
plt.plot(t, sig_fast, 'k', label='Original (Fast)')
plt.step(t, dm_overload, 'b', where='post', label=f'DM Staircase (Δ={delta_small})')
plt.title('Delta Modulation: Slope Overload')
plt.legend()
plt.grid(True)

# --- Visualization 3: MSE versus Step Size (Required visualization) ---
step_sizes = np.linspace(0.01, 0.4, 40)
mse_slow = []
mse_fast = []

for delta in step_sizes:
    dm_slow_test = simulate_delta_modulation(sig_slow, delta)
    dm_fast_test = simulate_delta_modulation(sig_fast, delta)
    mse_slow.append(np.mean((sig_slow - dm_slow_test)**2))
    mse_fast.append(np.mean((sig_fast - dm_fast_test)**2))

plt.subplot(4, 1, 4)
plt.plot(step_sizes, mse_slow, 'g-o', label='MSE (Slow Signal)')
plt.plot(step_sizes, mse_fast, 'm-o', label='MSE (Fast Signal)')
plt.title('Delta Modulation: Mean Squared Error (MSE) vs Step Size (Δ)')
plt.xlabel('Step Size (Δ)')
plt.ylabel('MSE')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

# --- Mandatory Validation Check ---
print("--- Mandatory Validation ---")
diffs = np.diff(dm_granular)
# Because of floating point arithmetic, we check if the absolute differences are very close to Delta
is_valid = np.allclose(np.abs(diffs[diffs != 0]), delta_large)
print(f"Are all non-zero step changes exactly +Δ or -Δ for granular test? {is_valid}")
```
